SETUP_MODEL

```python
import os
import sys
import subprocess

def install_dependencies():
    """Instala patool si no está presente."""
    try:
        import patoolib
    except ImportError:
        print("Instalando dependencia para descompresión (patool)...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", "patool"])

def setup_model_files():
    install_dependencies()
    import patoolib

    # Configuración de rutas
    # Reemplazar __file__ con os.getcwd() para entornos interactivos como Colab
    base_dir = os.getcwd()
    models_dir = os.path.join(base_dir, "Models")

    # El primer archivo de la secuencia RAR
    first_part = os.path.join(models_dir, "best_model.part1.rar")
    output_model = os.path.join(models_dir, "best_model.pth")

    print(f"Buscando archivos segmentados en: {models_dir}")

    if os.path.exists(output_model):
        print("El modelo 'best_model.pth' ya existe. Omitiendo descompresión.")
        return

    if not os.path.exists(first_part):
        print(f"Error: No se encontró {first_part}. Asegúrate de que todas las partes estén en la carpeta Models.")
        return

    try:
        print("Iniciando unión y descompresión de partes RAR...")
        # patool detecta automáticamente las partes (.part2, .part3...) si están en la misma carpeta
        patoolib.extract_archive(first_part, outdir=models_dir)
        print(f"Éxito: Modelo reconstruido en {output_model}")

    except Exception as e:
        print(f"Error durante la descompresión: {e}")
        print("Asegúrate de tener instalado 'unrar' (Win) o 'rar/unrar' (Linux/Mac) en el sistema.")

if __name__ == "__main__":
    setup_model_files()
```
