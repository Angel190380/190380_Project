```python
# 1. Instalar librerías necesarias
!pip install rasterio matplotlib numpy

import rasterio
import matplotlib.pyplot as plt
import numpy as np
from rasterio.enums import Resampling

# 2. RUTA CORREGIDA (Basada en tu panel de archivos)
input_tiff = '/content/CEM_Mosaico_Corregido.tif'
output_png = '/content/visualizacion_mde.png'

print(f"Iniciando la conversión de: {input_tiff}...")

try:
    with rasterio.open(input_tiff) as src:
        # 3. Leer los datos remuestreados para optimizar memoria
        scale_factor = 10
        new_width = int(src.width / scale_factor)
        new_height = int(src.height / scale_factor)

        data = src.read(
            1,
            out_shape=(src.count, new_height, new_width),
            resampling=Resampling.bilinear
        )

        # 4. Crear la visualización técnica
        plt.figure(figsize=(12, 10))

        # El mapa 'terrain' ayuda a identificar las zonas bajas de Veracruz
        im = plt.imshow(data, cmap='terrain')
        plt.colorbar(im, label='Elevación (m sobre el nivel del mar)')
        plt.title('Modelo Digital de Elevación (CEM) - Vista Previa')
        plt.axis('off')

        # 5. Guardar como PNG
        plt.savefig(output_png, dpi=300, bbox_inches='tight')
        plt.close()

    print(f"✅ Conversión exitosa. Imagen guardada en: {output_png}")

except Exception as e:
    # Si sigue fallando, esto imprimirá la causa exacta
    print(f"❌ Error al procesar el archivo: {e}")
```
