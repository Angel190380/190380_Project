```python
import torch
from torch.utils.data import Dataset, DataLoader
import numpy as np
import os

# 1. Configuración de Rutas (E5 Datasets)
base_path_e5 = "/content/drive/MyDrive/DOCTORADO 2026/VISION ARTIFICIAL/E5/Datasets/Train"

class FloodDataset(Dataset):
    def __init__(self, x_path, y_paths):
        """
        x_path: Ruta al CEM normalizado
        y_paths: Lista de rutas a los Ground Truth (Original, FlipH, FlipV)
        """
        self.X = np.load(x_path)
        # Cargamos y concatenamos las etiquetas aumentadas para tener más datos
        self.Y = [np.load(p) for p in y_paths]

        # Para este ejemplo, emparejamos el X original con cada aumento de Y
        # Nota: En un modelo real, X también debería sufrir el mismo aumento
        self.total_samples = len(self.Y)

    def __len__(self):
        return self.total_samples

    def __getitem__(self, idx):
        # Convertir a Tensores de PyTorch y agregar dimensión de canal (C, H, W)
        x_tensor = torch.from_numpy(self.X).float().unsqueeze(0)
        y_tensor = torch.from_numpy(self.Y[idx]).float().unsqueeze(0)

        return x_tensor, y_tensor

def inicializar_dataloader():
    # Definir archivos
    path_x = os.path.join(base_path_e5, "CEM_Normalized.npy")
    paths_y = [
        os.path.join(base_path_e5, "GT_Original.npy"),
        os.path.join(base_path_e5, "GT_FlipH.npy"),
        os.path.join(base_path_e5, "GT_FlipV.npy")
    ]

    # Crear instancia del Dataset
    dataset_veracruz = FloodDataset(path_x, paths_y)

    # Crear el DataLoader
    # batch_size=1 porque tus tensores de Veracruz son grandes
    train_loader = DataLoader(dataset_veracruz, batch_size=1, shuffle=True)

    print(f"DataLoader configurado con {len(dataset_veracruz)} muestras para entrenamiento.")

    # Prueba de carga de un batch
    for batch_x, batch_y in train_loader:
        print(f"Forma del tensor de entrada (X): {batch_x.shape}")
        print(f"Forma del tensor de salida (Y): {batch_y.shape}")
        break # Solo probamos el primero

    return train_loader

if __name__ == "__main__":
    loader = inicializar_dataloader()
```
