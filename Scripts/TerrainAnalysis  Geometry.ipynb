MAPA DE PENDIENTES

```python
from google.colab import drive
import os

# Montar el drive
drive.mount('/content/drive')

# --- CONFIGURA ESTA RUTA ---
# Ajusta 'Mi unidad/Tu_Carpeta/...' a la ubicación real de tu archivo en Drive
ruta_base = '/content/drive/MyDrive/DOCTORADO 2026/VISION ARTIFICIAL/E2/Terrain Model/'
input_path = os.path.join(ruta_base, 'Mosaico_Maestro_CEM_1.5m.tif')
output_path = os.path.join(ruta_base, 'Mapa_Pendientes_Veracruz.tif')

# Verificar si el archivo existe antes de continuar
if os.path.exists(input_path):
    print(f"Archivo encontrado: {input_path}")
else:
    print(f"ERROR: No se encontró el archivo en {input_path}. Verifica la ruta.")
```

```python
import os
from google.colab import drive
from osgeo import gdal


# 3. Generar el Mapa de Pendientes (Slope)
# 'slope' es la utilidad de GDAL para calcular la pendiente
# -format: Formato de salida (GeoTIFF)
# -p: Si se omite, calcula en grados. Si se incluye, calcula en porcentaje.
options = gdal.DEMProcessingOptions(format='GTiff', slopeFormat='degree')

try:
    gdal.DEMProcessing(output_path, input_path, 'slope', options=options)
    print(f"✅ ¡Éxito! Mapa de pendientes guardado en: {output_path}")
except Exception as e:
    print(f"❌ Error al procesar el archivo: {e}")
```

MODELO ORIENTACION

```python
import os
from osgeo import gdal

# 1. Definir la nueva ruta de salida
# Reutilizamos la variable 'ruta_base' que ya tienes configurada en tu notebook
output_path_aspect = os.path.join(ruta_base, 'Mapa_Orientacion_Aspect.tif')

print(f"Iniciando generación de Aspect para: {input_path}")

# 2. Configurar opciones para Aspect
# 'format': GeoTIFF para mantener la compatibilidad
options_aspect = gdal.DEMProcessingOptions(format='GTiff')

try:
    # 3. Ejecutar el procesamiento de Orientación (aspect)
    # El parámetro 'aspect' calcula la dirección de la pendiente en grados (0-360)
    gdal.DEMProcessing(output_path_aspect, input_path, 'aspect', options=options_aspect)
    print(f"✅ ¡Éxito! Mapa de orientación guardado en: {output_path_aspect}")
except Exception as e:
    print(f"❌ Error al procesar el archivo: {e}")
```

CAPA DE RUGOSIIDAD

```python
import os
from osgeo import gdal

# 1. Definir la ruta de salida para la Rugosidad
output_path_roughness = os.path.join(ruta_base, 'Capa_Rugosidad_Superficial.tif')

print(f"Iniciando generación de Rugosidad (TRI) para: {input_path}")

# 2. Configurar opciones para TRI (Terrain Ruggedness Index)
options_tri = gdal.DEMProcessingOptions(format='GTiff')

try:
    # 3. Ejecutar el procesamiento de Rugosidad
    # Usamos 'tri' para obtener el índice de rugosidad del terreno
    gdal.DEMProcessing(output_path_roughness, input_path, 'tri', options=options_tri)
    print(f"✅ ¡Éxito! Capa de rugosidad guardada en: {output_path_roughness}")
except Exception as e:
    print(f"❌ Error al procesar el archivo: {e}")
```
